## 4. Data limits

Unbalanced History: Newer queries or pages have shorter historical lookbacks compared to established ones, creating unequal sequence lengths.

GSC Sampling / Thresholding: Low-volume or long-tail search queries below Google Search Console thresholds are omitted, leading to an incomplete view of long-tail traffic.

Window Overlaps & Lag: Data updates may lag by 2–3 days, and rolling feature windows near the tail end of the time boundary risk data truncation.

Causal Inferences: Observed metrics reflect historical search engine visibility, not causal guarantees of ranking improvements.

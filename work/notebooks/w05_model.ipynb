## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

We implemented a GroupKFold split grouped by client/domain (or time-based split depending on your dataset structure).

Why this split is honest:

Prevents Data Leakage: Standard random splitting leaks information across rows originating from the same client or identical query sessions.

Simulates Real-World Evaluation: Grouping ensures the model is evaluated on completely unseen clients/entities during testing, accurately measuring how well it generalizes to new incoming data rather than memorizing historical entity-specific patterns.

```python
from sklearn.model_selection import GroupKFold
import numpy as np

# Example validation setup for GroupKFold
gkf = GroupKFold(n_splits=5)

# Verify group integrity check
# X: features, y: target, groups: client_id / domain array
print(f"Split strategy configured: GroupKFold with {gkf.get_n_splits()} splits.")
```

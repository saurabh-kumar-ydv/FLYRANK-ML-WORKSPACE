Unit of Analysis: One row represents a single unique URL aggregated over a 30-day performance window.

```python
import os
import pandas as pd

# Load dataset locally or download from starter repo
data_path = "content_refresh_anonymized.csv"
if not os.path.exists(data_path):
    !wget -q https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv

df = pd.read_csv(data_path)

# Preview columns to verify structure
print("Columns available:", df.columns.tolist())

# Calculate proxy drop percentage
if "clicks_prev" in df.columns and "clicks_curr" in df.columns:
    df["click_drop_pct"] = (df["clicks_prev"] - df["clicks_curr"]) / (df["clicks_prev"] + 1e-5)
elif "click_drop_pct" not in df.columns:
    df["click_drop_pct"] = (df["clicks"].max() - df["clicks"]) / (df["clicks"].max() + 1e-5)

# Create binary target column
df["needs_refresh"] = (df["click_drop_pct"] > 0.20).astype(int)

# Display unit of analysis preview
df.head()
```
